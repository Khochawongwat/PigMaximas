```python
import utils
import importlib
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

importlib.reload(utils)
img_path = "insert path here"
img = cv.imread(img_path)

threshold = 109
_, img = cv.threshold(img, threshold, 255, cv.THRESH_BINARY_INV)
img = img.astype(np.uint8)
img = cv.GaussianBlur(img, (11, 11), 0)
_, img = cv.threshold(img, 200, 255, cv.THRESH_BINARY)

sections = utils.section(img)

plt.figure(figsize=(12, 6))

plt.subplot(1, len(sections) + 1, 1)
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title("Image")
plt.axis("off")

for i, section in enumerate(sections):
    plt.subplot(1, len(sections) + 1, i + 2)
    plt.imshow(cv.cvtColor(section, cv.COLOR_BGR2RGB))
    plt.title(f"Section {i + 1}")
    plt.axis("off")

plt.tight_layout() 
plt.show()
```
